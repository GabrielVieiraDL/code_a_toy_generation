```python
while True:
  pergunta = input('Gostaria de jogar FizzBuzz?\n')
  if pergunta.lower() == 'sim' or pergunta.lower() == 's':
    numero = int(input('Informe um número maior que zero, para jogarmos: '))
    if numero <= 0:
      break
    else:
      for i in range(numero):
        if (i+1)%3 == 0 and (i+1)%5 == 0:
          print('FizzBuzz')
        elif (i+1)%3 == 0:
          print('Fizz')
        elif (i+1)%5 == 0:
          print('Buzz')
        else:
          print(i+1)
  else:
    break
```
