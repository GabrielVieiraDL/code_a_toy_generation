```python
# Para criar uma função em Python

# def nome_da_função(parâmetros):
  # sua função
  # return resultado
# Queremos criar uma função que é uma calculadora
def calculadora(num1, num2, operacao):

  if operacao == '+':
    return num1 + num2
  elif operacao == '-':
    return num1 + num2
  elif operacao == '*':
    return num1 * num2
  elif operacao == '/':
    if num2 == 0:
      return "Erro, não pode dividir por zero (0)"
    else:
      return num1 / num2
  else:
    return "Escolha um operador válido (+, -, * ou /)"
```

```python
# Vamos simular a nossa calculadora
num1 = float(input("Informe o primeiro número\n"))
num2 = float(input("Informe o segundo número\n"))
operacao = input("Informe um operador\n")

calculadora(num1, num2, operacao)
```

```python
# Agora vamos fazer uma calculadora conversora de moedas
# Vamos aprender sobre dicionários

dados_moedas = {
    'moeda' : ['Dólar', 'Euro', 'Libra'],
    'sigla' : ['USD', 'EUR', 'LIB'],
    'valor_reais' : [5.09, 5.86, 6.85]
}

print(dados_moedas)

# Se eu quero exibir apenas as chaves (colunas)
print(dados_moedas.keys())

# Se eu quero exibir apenas os valores (linhas)
print(dados_moedas.values())

# Quando queremos chamar uma "coluna" ou seja, chave apenas
print(dados_moedas['valor_reais'])

# Inserir nova coluna e valores
dados_moedas['pais'] = ['Estados Unidos da America', 'Europa', 'Reino Unido', 'Argentina']
print(dados_moedas)

# Exibir campo e posição específicos
print(dados_moedas['moeda'][2])

print('Bem-vindo ao seistema GenMoedas')
print('Informe qual moeda vocêr converter os seus milhões')
print('\n1 - Dólar\n2 - Euro\n3 - Libra\n4 - Informar nova moeda')

opcao = input('\nInforme a opção desejada\n')
valor_real = float(input('\nQual o valor quer converter?\n'))


if opcao == "1" or opcao.lower() == "dolar":
  dolar(valor_real)
elif opcao == "2" or opcao.lower() == "euro":
  euro(valor_real)
elif opcao == "3" or opcao.lower() == "libra":
  libra(valor_real)
elif opcao == "4" or opcao.lower() == "cadastrar":
  moeda = input('Informe o nome da moeda\n')
  sigla = input('Qual o valor da sigla\n')
  valor = float(input('Qual a cotação atual?\n'))
  cadastrar(moeda, sigla, valor)
  print(dados_moedas)
else:
  print("Opção não encontrada")
```

```python
def dolar(valor_real):
  return print(valor_real / 5.09)

def euro(valor_real):
  return print(valor_real / 5.86)

def libra(valor_real):
  return print(valor_real / 6.85)

def cadastrar(moeda, sigla, valor_real):
  #dict['chave'].append(parametro)
  dados_moedas['moeda'].append(moeda)
  dados_moedas['sigla'].append(sigla)
  dados_moedas['valor_reais'].append(valor_real)
```
