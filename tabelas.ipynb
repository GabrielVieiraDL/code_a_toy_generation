```python
# para um melhor relatório, uma tecnica que podemos fazer é unir bases de dados para um visual único

# queremos unir a base vendas com clientes

import pandas as pd

# vamos chamas as tabelas que queremos dar merge

vendas = '/content/dados/vendas.csv'
clientes = '/content/dados/clientes.csv'

df_vendas = pd.read_csv(vendas)
df_clientes = pd.read_csv(clientes)

#df_nome_tabela = pd.merge(tabela da esquerda, tabela da direita, left_on='nome do campo da tabela da esquerda', right_on='nome do campo da tabela da direita')
df_vendas_clientes = pd.merge(df_vendas, df_clientes, left_on='Cliente', right_on='Id Cliente', how = 'inner')

display(df_vendas_clientes)
```

```python
# queremos saber em uma única tabela o quanto nosso cliente gastou na conta

# ponto 1 - descobrir onde está o valor do produto
produtos = '/content/dados/produtos.csv'
df_produtos = pd.read_csv(produtos)
# ponto 2 - mesclar a tabela df_vendas_clientes com a tabela que onde está o valor
df_vendas_clientes_produtos = pd.merge(df_vendas_clientes, df_produtos, left_on='Produto', right_on='Produto', how= 'inner')
# ponto 3 - criar um novo campo, para mostrar o gasto na compra
df_vendas_clientes_produtos['Gasto'] = df_vendas_clientes_produtos['Quantidade'] * df_vendas_clientes_produtos['Valor']
# ponto 4 - display
display(df_vendas_clientes_produtos)
# ponto 5 - salvar o csv no computador
df_vendas_clientes_produtos.to_csv('Dados Completos.csv', index=False)
```
