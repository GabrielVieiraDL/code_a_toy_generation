```python
import pandas as pd

df_logs = pd.read_csv('/content/logs_seguranca_aula.csv')

display(df_logs.head(10))

# Sempre que temos um "timestamp" devemos verificar se ele é um timestamp
df_logs['timestamp'] = pd.to_datetime(df_logs['timestamp'])
df_logs.info()
df_logs.describe()
```

```python
# Descobrir logins realizados fora do horário comercial
# seg - sex - 08 - 17

# o python pode identificar dentro de um timestamp valores ramificados
df_logs['hora'] = df_logs['timestamp'].dt.hour
df_logs['minutos'] = df_logs['timestamp'].dt.minute
df_logs['dia_semana'] = df_logs['timestamp'].dt.day_of_week
df_logs['nome_dia'] = df_logs['timestamp'].dt.day_name()

# Criar os filtros
# Fora do horário comercial das 00h até as 07h e das 17h até as 23h
display(df_logs[df_logs['hora'] != 8])

# Vamos exibir os dados que não são resource financeiro ou system

# or -----> dentro de if, for ou while
# | ------> operador de OU
# and -----> dentro de if, for ou while
# & -------> operador de E

display(df_logs[(df_logs['resource'] != 'financeiro') & (df_logs['resource'] != 'system')])

# filtrar colunas diferentes
# Somente falhas na sexta-feira

display(df_logs[(df_logs['status'] == "FAILED") & (df_logs['dia_semana'] == 6)])

# Encontrar informações que iniciam ou finalizam com
# startswith - é inicia com
# endswith - finaliza com
# mostrar tudo que termina com .csv

display(df_logs[df_logs['resource'].str.endswith(".csv", na=False)])
```
