```python
import requests
import json

endpoint_sheets = "https://script.google.com/macros/s/AKfycbwZUWP_qC88ENPWWsbYQjbVzpKmg3KXTD7ymlCtLLRZIAm6y5sLUqrALHUeAKTgnpf7sA/exec"

# post - postar - colocar
# get - pegar - quando eu coleto uma informação de algum lugar

def enviar_msg(nome, email):
  response = requests.post(
      #onde
      endpoint_sheets,
      #o que
      json={"nome":nome, "email":email}
  )
  # para enviar uma msg do python para o sheets, devemos chamar o metodo request.post(onde, o que)
  # se quiser ver a mensagem enviada
  print(response.text)
```

```python
nome = input("Informe o nome desejado:\n")
email = input("Informe o email desejado\n")

enviar_msg(nome, email)
```
