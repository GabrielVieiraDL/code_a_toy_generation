```python
import pandas as pd

dados_musica = "/content/playlist_hits.json"
df_dados_musica = pd.read_json(dados_musica)



list_tamanho_musica = []
list_duracoes = df_dados_musica["duracao_seg"].tolist()
#.tolist() faz a colunar virar uma lista python

#se musica maior que ou igual a 300,salvar na list_tamanho_musica
#e se menor que 300 salvar como "musica_curta"

for i in list_duracoes:
  if i >= 300:
    list_tamanho_musica.append("musica_grande")
  else:
    list_tamanho_musica.append("musica_curta")

display(list_tamanho_musica)
df_dados_musica["categoria_tamanho"] = list_tamanho_musica
display(df_dados_musica)
```

```python
#ver metricas
media_stream = df_dados_musica["streams_milhoes"].mean()
print(media_stream)
```

```python
df_dados_musica_pop = df_dados_musica[df_dados_musica["genero"] == "Pop"]
display(df_dados_musica_pop)
print(f"o total de streams é de {df_dados_musica["streams_milhoes"].sum()}")
print("-"*90)
df_dados_musica_rock = df_dados_musica[df_dados_musica["genero"] == "Rock"]
display(df_dados_musica_rock)
print(f"a média de streams de rock é {df_dados_musica_rock["streams_milhoes"].mean()}")
```

```python
#filtro simples, mas com dois criterios
#metodo do .isin()

filtro_trilha_pop = ["Pop","Trilha Sonora"]
df_dados_musica_trilha_pop = df_dados_musica[df_dados_musica["genero"].isin(filtro_trilha_pop)]
display(df_dados_musica_trilha_pop)

```
