```python
# Agora queremos que o usuário informe todos os valores de uma comanda
# Laço - while (enquanto) - ter condoção ou não
# Se não tem condição EU TENHO QUE FALAR QUANDO TERMINA

saldo = 0

while True:
  # while funciona enquanto a condição for verdadeira
  valor = float(input('Informe o valor do item: '))
  saldo = saldo + valor

  pergunta = input('Tem um novo valor a informar? ')

  if pergunta.lower() == "sim" or pergunta.lower() == 's':
    print('Informe novo valor: ')
  else:
    print(f'O seu saldo é de {saldo}')
    break # break é o comando que finaliza o while true
```

```python
# quero casdastrar o produto consumido
list_produtos = []

while True:
  produto = input('Informe o nome do produto: \n')

  # Como colocar o valor dentro da lista?
  list_produtos.append(produto)

  pergunta = input('Tem um novo produto a informar? \n')

  if pergunta.lower() == "sim" or pergunta.lower() == 's':
    print('Informe novo valor: ')
  else:
   #print(list_produtos) - Não vamos exibir a lista assim.
   # O comando FOR percorre uma estrutura.
   # for VARIAVEL in ESTRUTURA
   for i in list_produtos:
    print(i)
    break # break é o comando que finaliza o while true
```

```python
# Agora vamos cadastrar o produto e seu preço
#depois vamos somar o total da compra

list_produtos = []
list_valores = []

# --------- Cadastro das Informações
while True:
  produto = input('Informe o nome do produto:\n')
  valor = float(input(f'Informe o valor para {produto}\n'))

  list_produtos.append(produto)
  list_valores.append(valor)

  pergunta = input('Deseja cadastrar um novo produto?\n')

  if pergunta.lower() == 'sim' or pergunta.lower() == 's':
    print('Vamos iniciar novo cadastro')
  else:
    # i - produto
    # j - valor
    for i, j in zip(list_produtos, list_valores):
      # produto -- R$ valor
      print(f'O produto {i} custa R$ {j}')

    print(f'\n\n\n Total da compra {sum(list_valores)}')
    # O comando sum(LIST) soma todos os valores de dentro de uma lista
    # O comandao sum(LISTA, VALOR) faz com que ele some esse valor que informamos também.
    taxa = sum(list_valores)*0.1
    print(f'Valor com taxa de serviço {round(sum(list_valores,taxa),2)}')
    break
```
