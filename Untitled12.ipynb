```python
import hashlib #biblioteca de hashs
#hash - é um método de criptografia """"padrão""""

senha = 'generation'

# método sha256
senha_encode = senha.encode("utf-8")

senha_hash = hashlib.sha256(senha_encode).hexdigest()

print(senha_hash)
```
